# Introduction à QuTiP

Ce notebook montre comment utiliser **QuTiP** (Quantum Toolbox in Python) pour manipuler des états quantiques, des opérateurs, et simuler l’évolution temporelle d’un système simple.

Il fonctionne directement dans Binder via le bouton “marche”.


```python
import numpy as np
import matplotlib.pyplot as plt
import qutip as qt

plt.rcParams["figure.figsize"] = (6,4)
```

```python
# |0> et |1>
ket0 = qt.basis(2, 0)
ket1 = qt.basis(2, 1)

ket0, ket1
sx = qt.sigmax()
sy = qt.sigmay()
sz = qt.sigmaz()

sx, sy, sz

# Action de σx sur |0>
result = sx * ket0
result



```

```python
b = qt.Bloch()
b.add_states([ket0, ket1, result])
b.show()
```

```python
# Hamiltonien : H = (ω/2) σz
omega = 1.0
H = 0.5 * omega * sz

# état initial : |0>
psi0 = ket0

# temps
tlist = np.linspace(0, 10, 200)

# résolution de l’équation de Schrödinger
result = qt.sesolve(H, psi0, tlist)

# probabilité d’être dans |1>
p1 = [abs((ket1.dag() * psi)[0][0])**2 for psi in result.states]

plt.plot(tlist, p1)
plt.xlabel("t")
plt.ylabel("Probabilité d'être dans |1>")
plt.title("Oscillations induites par H = (ω/2) σz")
plt.grid(True)
plt.show()
```
